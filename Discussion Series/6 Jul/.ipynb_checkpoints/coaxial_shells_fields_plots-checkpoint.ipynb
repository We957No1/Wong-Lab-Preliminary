# Two coaxial long cylindrical shells: field profiles and graphics

This notebook accompanies the LaTeX derivation. The default model is the canonical coaxial pair:

- inner shell radius `a`, outer shell radius `b`
- inner shell charge per unit length `+lambda_per_length`, outer shell `-lambda_per_length`
- inner shell axial current `+I0`, outer shell axial current `-I0`
- homogeneous annular material with permittivity `eps` and permeability `mu`

The static fields are confined to the annular region `a < r < b`. The electrostatic fields are radial, the magnetostatic fields are circumferential, and a convenient vector potential is axial.

```python
import numpy as np
import matplotlib.pyplot as plt

# Physical constants
_eps0 = 8.8541878128e-12
_mu0 = 4*np.pi*1e-7

# Geometry: two coaxial cylindrical shells
# You can change these values and rerun the notebook.
a = 1.0e-2      # inner radius, m
b = 5.0e-2      # outer radius, m

# Material in the annulus
eps_r = 1.0
mu_r = 1.0
eps = eps_r * _eps0
mu = mu_r * _mu0

# Source strengths for the canonical coaxial pair
lambda_per_length = 1.0e-9   # C/m on inner shell; outer shell has the negative
I0 = 1.0                     # A on inner shell; outer shell has the return current

# Useful derived constants
log_ba = np.log(b/a)
V0 = lambda_per_length/(2*np.pi*eps) * log_ba
Cprime = 2*np.pi*eps/log_ba
Lprime = mu/(2*np.pi)*log_ba
Z0 = np.sqrt(Lprime/Cprime)
v_phase = 1/np.sqrt(Lprime*Cprime)

print(f"a = {a:.4g} m, b = {b:.4g} m")
print(f"V0 = V(a)-V(b) = {V0:.6g} V")
print(f"C' = {Cprime:.6e} F/m")
print(f"L' = {Lprime:.6e} H/m")
print(f"Z0 = sqrt(L'/C') = {Z0:.6g} ohm")
print(f"1/sqrt(L'C') = {v_phase:.6e} m/s")
```

## Helper functions

The functions below use the piecewise canonical coaxial solution:

- `electrostatic_profiles(r)` returns `D_r`, `E_r`, and `V`
- `magnetostatic_profiles(r)` returns `H_phi`, `B_phi`, and `A_z`
- `mqs_profiles(r,t)` returns the slowly time-varying fields for a sinusoidal current

```python
def annulus(r):
    r = np.asarray(r)
    return (r >= a) & (r <= b)


def electrostatic_profiles(r, lam=lambda_per_length):
    # Return D_r, E_r, V for inner charge +lam and outer charge -lam.
    r = np.asarray(r, dtype=float)
    D = np.zeros_like(r)
    E = np.zeros_like(r)
    V = np.zeros_like(r)

    mask = annulus(r)
    inside = r < a

    D[mask] = lam/(2*np.pi*r[mask])
    E[mask] = D[mask]/eps
    V[mask] = lam/(2*np.pi*eps) * np.log(b/r[mask])
    V[inside] = lam/(2*np.pi*eps) * log_ba
    # V is set to zero outside r>b.
    return D, E, V


def magnetostatic_profiles(r, current=I0):
    # Return H_phi, B_phi, A_z for inner current +current and outer return current -current.
    r = np.asarray(r, dtype=float)
    H = np.zeros_like(r)
    B = np.zeros_like(r)
    A = np.zeros_like(r)

    mask = annulus(r)
    inside = r < a

    H[mask] = current/(2*np.pi*r[mask])
    B[mask] = mu*H[mask]
    A[mask] = mu*current/(2*np.pi) * np.log(b/r[mask])
    A[inside] = mu*current/(2*np.pi) * log_ba
    # Gauge choice: A_z = 0 for r>b.
    return H, B, A


def mqs_profiles(r, t, I_amplitude=I0, V_amplitude=V0, frequency=1.0e5):
    # Magnetoquasistatic snapshot.
    # I(t) = I_amplitude*sin(omega*t) and V0(t) = V_amplitude*cos(omega*t).
    # Returns radial electrostatic fields, axial inductive fields,
    # circumferential magnetic fields, and the potentials.
    r = np.asarray(r, dtype=float)
    omega = 2*np.pi*frequency
    It = I_amplitude*np.sin(omega*t)
    dIdt = I_amplitude*omega*np.cos(omega*t)
    Vt = V_amplitude*np.cos(omega*t)

    mask = annulus(r)
    inside = r < a

    Er = np.zeros_like(r)
    Dr = np.zeros_like(r)
    V = np.zeros_like(r)
    Ez_ind = np.zeros_like(r)
    Dz_ind = np.zeros_like(r)

    Er[mask] = Vt/(r[mask]*log_ba)
    Dr[mask] = eps*Er[mask]
    V[mask] = Vt*np.log(b/r[mask])/log_ba
    V[inside] = Vt

    # Gauge A_z(b,t)=0. Then E_z_ind = -partial A_z/partial t in the annulus.
    Ez_ind[mask] = -mu*dIdt/(2*np.pi)*np.log(b/r[mask])
    Dz_ind[mask] = eps*Ez_ind[mask]

    H, B, A = magnetostatic_profiles(r, current=It)

    return {
        "t": t,
        "I": It,
        "dIdt": dIdt,
        "V0": Vt,
        "Er": Er,
        "Dr": Dr,
        "V": V,
        "Ez_ind": Ez_ind,
        "Dz_ind": Dz_ind,
        "Hphi": H,
        "Bphi": B,
        "Az": A,
    }


def plot_profile(r, y, ylabel, title):
    plt.figure(figsize=(7.2, 4.2))
    plt.plot(r, y)
    plt.axvline(a, linestyle="--", linewidth=1)
    plt.axvline(b, linestyle="--", linewidth=1)
    plt.xlabel("radius r (m)")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.show()
```

## 1. Electrostatic profiles: D, E, and V

For the canonical pair, only the annulus has nonzero electric field. The potential is constant inside the inner shell and zero outside the outer shell under the gauge choice `V(b)=0`.

```python
r = np.linspace(0.2*a, 1.8*b, 1500)
D_r, E_r, V = electrostatic_profiles(r)

plot_profile(r, D_r, "D_r (C/m^2)", "Electrostatics: radial displacement field D_r")
plot_profile(r, E_r, "E_r (V/m)", "Electrostatics: radial electric field E_r")
plot_profile(r, V, "V (V)", "Electrostatics: scalar potential V with V(b)=0")
```

## 2. Magnetostatic profiles: H, B, and A

The current is axial, so the magnetic field is circumferential. The vector potential is chosen as `A = A_z(r) zhat`, with gauge `A_z(b)=0` and `A_z=0` outside the outer shell.

```python
H_phi, B_phi, A_z = magnetostatic_profiles(r)

plot_profile(r, H_phi, "H_phi (A/m)", "Magnetostatics: circumferential H_phi")
plot_profile(r, B_phi, "B_phi (T)", "Magnetostatics: circumferential B_phi")
plot_profile(r, A_z, "A_z (T m)", "Magnetostatics: axial vector potential A_z")
```

## 3. Magnetoquasistatic profiles: slowly varying current

Here the current and voltage are taken as slowly varying sinusoidal functions. The magnetic field is still obtained instantaneously from Ampere's law, while Faraday induction adds an axial electric field.

The chosen gauge gives `E_z_ind = -partial A_z/partial t`. Its radial derivative is fixed by Faraday's law.

```python
frequency = 1.0e5  # Hz; still quasistatic for centimeter-scale geometry in vacuum
# Choose a time where both I(t) and dI/dt are nonzero.
t_snapshot = 1/(8*frequency)
fields_t = mqs_profiles(r, t_snapshot, frequency=frequency)

print(f"snapshot time t = {t_snapshot:.6e} s")
print(f"I(t) = {fields_t['I']:.6g} A")
print(f"dI/dt = {fields_t['dIdt']:.6e} A/s")
print(f"V0(t) = {fields_t['V0']:.6g} V")

plot_profile(r, fields_t["Er"], "E_r (V/m)", "MQS snapshot: radial electric field from scalar potential")
plot_profile(r, fields_t["Ez_ind"], "E_z_ind (V/m)", "MQS snapshot: axial inductive electric field")
plot_profile(r, fields_t["Bphi"], "B_phi (T)", "MQS snapshot: instantaneous circumferential magnetic field")
plot_profile(r, fields_t["Az"], "A_z (T m)", "MQS snapshot: vector potential A_z")
```

## 4. Interesting extension: Poynting-vector energy flow

When the coaxial pair has both a voltage and a current, the static-looking fields imply a nonzero Poynting vector. In the annulus,

`S_z(r) = E_r(r) H_phi(r) = V0*I/(2*pi*r^2*log(b/a))`.

Integrating over the annular cross-section should give the circuit power `P = V0*I`.

```python
def poynting_profile(r, voltage=V0, current=I0):
    r = np.asarray(r, dtype=float)
    S = np.zeros_like(r)
    mask = annulus(r)
    S[mask] = voltage*current/(2*np.pi*r[mask]**2*log_ba)
    return S

S_z = poynting_profile(r)

# Use an integration grid strictly inside the annulus for better accuracy.
r_int = np.linspace(a, b, 50000)
S_int = poynting_profile(r_int)
P_numeric = np.trapezoid(S_int*2*np.pi*r_int, r_int)
P_exact = V0*I0

print(f"Numerical integral of S_z over annulus = {P_numeric:.8g} W")
print(f"Circuit power V0*I0 = {P_exact:.8g} W")
print(f"relative error = {(P_numeric-P_exact)/P_exact:.3e}")

plot_profile(r, S_z, "S_z (W/m^2)", "Poynting vector: axial energy-flow density S_z")
```

## 5. Energy density check

The electric and magnetic energy per unit length should satisfy

- `U_E' = 0.5*Cprime*V0^2`
- `U_B' = 0.5*Lprime*I0^2`

```python
u_E = 0.5*eps*E_r**2
u_B = 0.5*mu*H_phi**2

# Use a dense grid in the annulus for accurate energy integrals.
D_int, E_int, V_int = electrostatic_profiles(r_int)
H_int, B_int, A_int = magnetostatic_profiles(r_int)
u_E_int = 0.5*eps*E_int**2
u_B_int = 0.5*mu*H_int**2

U_E_numeric = np.trapezoid(u_E_int*2*np.pi*r_int, r_int)
U_B_numeric = np.trapezoid(u_B_int*2*np.pi*r_int, r_int)
U_E_exact = 0.5*Cprime*V0**2
U_B_exact = 0.5*Lprime*I0**2

print(f"U_E' numerical = {U_E_numeric:.8e} J/m")
print(f"U_E' formula   = {U_E_exact:.8e} J/m")
print(f"U_B' numerical = {U_B_numeric:.8e} J/m")
print(f"U_B' formula   = {U_B_exact:.8e} J/m")

plot_profile(r, u_E, "u_E (J/m^3)", "Electric energy density")
plot_profile(r, u_B, "u_B (J/m^3)", "Magnetic energy density")
```

## 6. Cross-sectional graphics

The first graphic shows the axial Poynting-vector density in the annulus. The second and third graphics show the field directions in the transverse plane.

```python
# Poynting-vector heatmap in the cross-section
N = 300
extent = 1.2*b
x = np.linspace(-extent, extent, N)
y = np.linspace(-extent, extent, N)
X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
mask = (R >= a) & (R <= b)

S_map = np.full_like(R, np.nan)
S_map[mask] = V0*I0/(2*np.pi*R[mask]**2*log_ba)

plt.figure(figsize=(6.2, 5.2))
im = plt.imshow(S_map, extent=[-extent, extent, -extent, extent], origin="lower")
plt.gca().set_aspect("equal")
plt.colorbar(im, label="S_z (W/m^2)")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Cross-section: axial Poynting-vector density")
plt.show()
```

```python
# Direction field for the radial electric field
Nq = 23
xq = np.linspace(-extent, extent, Nq)
yq = np.linspace(-extent, extent, Nq)
Xq, Yq = np.meshgrid(xq, yq)
Rq = np.sqrt(Xq**2 + Yq**2)
maskq = (Rq >= a) & (Rq <= b)

Ex_dir = np.zeros_like(Xq)
Ey_dir = np.zeros_like(Yq)
Ex_dir[maskq] = Xq[maskq]/Rq[maskq]
Ey_dir[maskq] = Yq[maskq]/Rq[maskq]

plt.figure(figsize=(6.2, 5.2))
plt.quiver(Xq[maskq], Yq[maskq], Ex_dir[maskq], Ey_dir[maskq])
plt.gca().set_aspect("equal")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Direction field: radial electric field")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Direction field for the circumferential magnetic field
Hx_dir = np.zeros_like(Xq)
Hy_dir = np.zeros_like(Yq)
Hx_dir[maskq] = -Yq[maskq]/Rq[maskq]
Hy_dir[maskq] = Xq[maskq]/Rq[maskq]

plt.figure(figsize=(6.2, 5.2))
plt.quiver(Xq[maskq], Yq[maskq], Hx_dir[maskq], Hy_dir[maskq])
plt.gca().set_aspect("equal")
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Direction field: circumferential magnetic field")
plt.grid(True, alpha=0.3)
plt.show()
```

## 7. Optional: change the geometry and sources

Try changing `a`, `b`, `lambda_per_length`, `I0`, `eps_r`, or `mu_r` in the first code cell and rerun the notebook. The most important qualitative changes are:

- larger `b/a` increases both the voltage for a fixed charge per length and the inductance per length
- the capacitance per length decreases as `log(b/a)` increases
- the Poynting-vector density always peaks near the inner conductor because it scales as `1/r^2`
